```python
def profile_dataset_with_json(df):
    metadata = datamart_profiler.process_dataset(df)
    result = {}

    for column_meta in metadata['columns']:
        name = column_meta['name']
        col_data = df[name]
        sample_val = col_data.dropna().iloc[0]
        if isinstance(sample_val, np.integer):
          sample_val = int(sample_val)

        stats = {
            "dtype": str(col_data.dtype),
            "num_nulls": int(col_data.isnull().sum()),
            "num_unique": int(col_data.nunique(dropna=True)),
            "sample_value": sample_val,
            "structural_type": column_meta.get('structural_type', 'Unknown').split('/')[-1],
            "coverage": None,
        }

        # Coverage (numerical range if exists)
        if 'coverage' in column_meta:
            ranges = [c['range'] for c in column_meta['coverage']]
            stats['coverage'] = {
                "min": min([r['gte'] for r in ranges]),
                "max": max([r['lte'] for r in ranges])
            }

        result[name] = stats

    # Temporal
    if 'temporal_coverage' in metadata:
        result["_temporal"] = []
        for temp_cov in metadata['temporal_coverage']:
            col_names = temp_cov['column_names']
            resolution = temp_cov['temporal_resolution']
            range_values = [entry['range']['gte'] for entry in temp_cov['ranges']] + [entry['range']['lte'] for entry in temp_cov['ranges']]
            result["_temporal"].append({
                "columns": col_names,
                "temporal_resolution": resolution,
                "min_date": datetime.fromtimestamp(min(range_values)).strftime("%Y-%m-%d"),
                "max_date": datetime.fromtimestamp(max(range_values)).strftime("%Y-%m-%d")
            })

    # Spatial
    if 'spatial_coverage' in metadata:
        result["_spatial"] = []
        for spatial_cov in metadata['spatial_coverage']:
            result["_spatial"].append({
                "columns": spatial_cov['column_names'],
                "spatial_type": spatial_cov['type']
            })

    original_column_names = list(df.columns)
    return result, original_column_names
```

```python
%%writefile app.py

import streamlit as st
import pandas as pd
import json
from pydantic import BaseModel
from typing import List
import tempfile
from google import genai

client = genai.Client(api_key="AIzaSyB2vgqd9To32UvfnEeWUtDeeoJUIhLGHy0")

# schema
class Relationship(BaseModel):
    from_column: str
    to_table: str
    to_column: str
    reason: str

class Column(BaseModel):
    name: str
    type: str
    description: str
    example_value: List[str]

class Metadata(BaseModel):
    table_name: str
    description: str
    tags: List[str]
    columns: List[Column]
    relationships: List[Relationship] = []

class MetadataNoRel(BaseModel):
    table_name: str
    description: str
    tags: List[str]
    columns: List[Column]

class Data(BaseModel):
    database_name: str
    metadata: List[Metadata]

N_SAMPLE_ROWS = 5

#call
def call_llm_2(prompt):
  response = client.models.generate_content(
      model="gemini-2.0-flash",
      contents=prompt,
      config={
          "response_mime_type": "application/json",
          "response_schema": MetadataNoRel,
      },
  )
  return response.text

def enrich_metadata_with_relationships(all_metadata: list[Metadata]):
    prompt = generate_enrichment_prompt(all_metadata)
    response = client.models.generate_content(
        model="gemini-2.0-flash",
        contents=prompt,
        config={
            "response_mime_type": "application/json",
            "response_schema": Data,

        },
    )
    return response.text


# prompt
def generate_prompt(table_name, sample_data, columns):
    return f"""
You are a data catalog assistant. Your task is to generate metadata for a dataset.

Table name: "{table_name}"

Sample data:
{json.dumps(sample_data, indent=2)}

Columns:
{columns}

For each column, provide:
- name
- type (e.g., string, integer, float, date, boolean)
- description
- Example values from the sample
Also provide a table-level description and relevant tags. If it might be a primary or foreign key, indicate so in the description.

Output the metadata as a JSON object.
"""

def generate_enrichment_prompt(metadata_list: list[Metadata]) -> str:
    metadata_jsons = [m.model_dump() for m in metadata_list]
    return f"""
You are a metadata enrichment assistant. Your task is to analyze multiple tables' metadata and enrich each with possible relationships to other tables.

Here is the current metadata for all tables:
{json.dumps(metadata_jsons, indent=2)}

Instructions:
- For each table, look at all columns and try to identify if any of them reference another table's primary key.
- Add a "relationships" key to each table, with a list of relationship objects.


Return the full metadata for the database.
"""

# st
st.title("Metadata Generator")

uploaded_files = st.file_uploader("Upload CSV files", type="csv", accept_multiple_files=True)

if uploaded_files:
    all_metadata = []

    for uploaded_file in uploaded_files:
        table_name = uploaded_file.name.split('.')[0]
        df = pd.read_csv(uploaded_file)
        st.subheader(f"{table_name}")


        sample_data = (
            df.drop_duplicates()
            .sample(n=N_SAMPLE_ROWS, random_state=42)
            .to_dict(orient="records")
        )
        st.dataframe(sample_data)
        columns = list(df.columns)

        prompt = generate_prompt(table_name, sample_data, columns)
        llm_response = call_llm_2(prompt)

        try:
          metadata_no_rel_obj = MetadataNoRel.model_validate_json(llm_response)
          metadata_obj = Metadata(**metadata_no_rel_obj.model_dump())
          all_metadata.append(metadata_obj)

          st.markdown(f"### Description for `{table_name}`")
          st.write(metadata_obj.description)

          columns_df = pd.DataFrame([col.model_dump() for col in metadata_obj.columns])
          columns_df = columns_df[["name", "type", "description", "example_value"]]

          st.markdown("### Column Metadata")
          st.dataframe(columns_df, use_container_width=True)

          with st.expander("Show Raw JSON"):
            st.json(metadata_obj.model_dump())

          # st.markdown("### Tags")
          # selected_tags = st.multiselect("Filter columns by tag", metadata_obj.tags)
          # if selected_tags:
          #   st.write("Filtering by tags:", selected_tags)
          #   filtered_columns = [col for col in metadata_obj.columns if any(tag.lower() in col.description.lower() for tag in selected_tags)]
          #   filtered_df = pd.DataFrame([col.model_dump() for col in filtered_columns])
          #   st.dataframe(filtered_df[["name", "type", "description", "example_value"]], use_container_width=True)

          st.markdown("### Export")
          metadata_json = json.dumps(metadata_obj.model_dump(), indent=2)
          st.download_button("Download JSON", metadata_json, file_name=f"{table_name}_metadata.json", mime="application/json")

          # metadata_csv = columns_df.to_csv(index=False)
          # st.download_button("download CSV (columns Only)", metadata_csv, file_name=f"{table_name}_columns.csv", mime="text/csv")
        except Exception as e:
          st.error(f"Failed to parse metadata: {e}")



    # Optional enrichment (Pass 2)
    if st.button("Enrich with Relationships"):
        enrichment_prompt = json.dumps([m.model_dump() for m in all_metadata], indent=2)
        enriched_response = enrich_metadata_with_relationships(all_metadata)

        try:
            data_obj = Data.model_validate_json(enriched_response)
            st.subheader("Enriched Metadata")
            st.json(data_obj.model_dump())
        except Exception as e:
            st.error(f"Enrichment failed: {e}")


```
